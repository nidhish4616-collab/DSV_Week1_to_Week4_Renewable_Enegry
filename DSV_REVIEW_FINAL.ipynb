Step 4: Remove Completely Empty Column

```python
# Remove completely empty columns
df.dropna(axis=1, how='all', inplace=True)

# Numerical missing values
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Categorical missing values
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    df[col] = df[col].fillna("Unknown")

# Remove duplicates
df.drop_duplicates(inplace=True)

print("Total Missing Values:", df.isnull().sum().sum())
```

Step 5: Z-Score Outlier Detection

```python
from scipy.stats import zscore

numeric_cols = df.select_dtypes(include=np.number).columns

z = np.abs(zscore(df[numeric_cols], nan_policy='omit'))

print("Z-score outliers (|Z| > 3):")

display(pd.Series((z > 3).sum(axis=0), index=numeric_cols))
```

Step 6: IQR Outlier Detection

```python
Q1 = df[numeric_cols].quantile(0.25)

Q3 = df[numeric_cols].quantile(0.75)

IQR = Q3 - Q1

outliers = ((df[numeric_cols] < Q1 - 1.5 * IQR) |
            (df[numeric_cols] > Q3 + 1.5 * IQR))

print("IQR Outliers:")

display(outliers.sum())
```

Step 7: IQR Outlier Detection

```python
plt.figure(figsize=(10,5))

sns.boxplot(data=df[numeric_cols])

plt.title("Outlier Detection using Boxplot")

plt.show()
```

# Week 6 – Feature Engineering

## Step 1: Import libraries

```python
import io
import zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
from sklearn.preprocessing import MinMaxScaler, StandardScaler, LabelEncoder

sns.set_theme(style="whitegrid")
print("Libraries imported successfully.")
```

## Step 2: Load the new dataset

Upload `NewsData.io_Renewable_Energy.csv (1).zip` or the extracted CSV file. This cell reads either format and creates the DataFrame `df`.

```python
uploaded = files.upload()
filename = next(iter(uploaded))
file_bytes = uploaded[filename]

if filename.lower().endswith(".zip"):
    with zipfile.ZipFile(io.BytesIO(file_bytes)) as z:
        csv_files = [name for name in z.namelist() if name.lower().endswith(".csv")]
        if not csv_files:
            raise ValueError("The ZIP file does not contain a CSV file.")
        with z.open(csv_files[0]) as f:
            df = pd.read_csv(f)
elif filename.lower().endswith(".csv"):
    df = pd.read_csv(io.BytesIO(file_bytes))
else:
    raise ValueError("Please upload a ZIP or CSV file.")

print("Dataset loaded successfully.")
print("File:", filename)
print("Dataset shape:", df.shape)
display(df.head())
```

## Step 3: Inspect dataset columns

Check that the columns required for Week 6 feature engineering are present.

```python
print("Dataset columns:")
print(df.columns.tolist())

required_columns = [
    "title", "description", "content", "keywords",
    "pubDate", "source_priority", "category", "sentiment", "language"
]

missing_columns = [col for col in required_columns if col not in df.columns]
if missing_columns:
    raise ValueError("Required columns missing from dataset: " + str(missing_columns))

print("\nAll required columns are available.")
```

## Step 4: Create new features

Create article title length, description length, content length, keyword count, publication year and month. These features convert text and date information into useful numerical variables.

```python
df_week6 = df.copy()

df_week6["Title_Length"] = df_week6["title"].fillna("").astype(str).str.len()
df_week6["Description_Length"] = df_week6["description"].fillna("").astype(str).str.len()
df_week6["Content_Length"] = df_week6["content"].fillna("").astype(str).str.len()

df_week6["Keyword_Count"] = df_week6["keywords"].fillna("").astype(str).apply(
    lambda x: 0 if x.strip() == "" else len([v for v in x.split(",") if v.strip()])
)

date = pd.to_datetime(df_week6["pubDate"], errors="coerce", utc=True)
df_week6["Year"] = date.dt.year
df_week6["Month"] = date.dt.month
df_week6["source_priority"] = pd.to_numeric(
    df_week6["source_priority"], errors="coerce"
)

new_features = [
    "Title_Length", "Description_Length", "Content_Length",
    "Keyword_Count", "Year", "Month", "source_priority"
]
display(df_week6[new_features].head())
print("New numerical features created.")
```

## Step 5: Prepare numerical features

Select the numerical features, replace infinite values and fill missing values using the median. This prepares the data for statistical analysis and scaling.

```python
feature_columns = [
    "source_priority", "Year", "Month", "Title_Length",
    "Description_Length", "Content_Length", "Keyword_Count"
]

X = df_week6[feature_columns].replace([np.inf, -np.inf], np.nan).copy()
X = X.fillna(X.median()).fillna(0)

print("Numerical features:")
print(feature_columns)
display(X.describe().round(2))
```

## Step 6: Calculate skewness

Skewness measures how asymmetric a numerical feature's distribution is. A value near zero indicates a more symmetric distribution.

```python
print("Skewness before transformation:")
display(X.skew().to_frame(name="Skewness").round(3))
```

## Step 7: Visualize distribution before transformation

The histogram and KDE curve show the original distribution of article content length.

```python
plt.figure(figsize=(9, 5))
sns.histplot(X["Content_Length"], kde=True)
plt.title("Content Length Before Transformation")
plt.xlabel("Content Length")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

## Step 8: Apply skewness correction

Use `log1p()` on non-negative numerical features to reduce positive skewness. The original columns are kept unchanged; transformed values are stored in new columns.

```python
skew_features = [
    "Title_Length", "Description_Length", "Content_Length", "Keyword_Count"
]

for col in skew_features:
    df_week6[col + "_Log"] = np.log1p(X[col].clip(lower=0))

print("Skewness after log transformation:")
display(df_week6[[col + "_Log" for col in skew_features]].skew().to_frame(
    name="Skewness"
).round(3))
```

## Step 9: Visualize distribution after transformation

Compare the content-length distribution after log transformation.

```python
plt.figure(figsize=(9, 5))
sns.histplot(df_week6["Content_Length_Log"], kde=True)
plt.title("Content Length After Log Transformation")
plt.xlabel("Log Transformed Content Length")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

## Step 10: One-Hot Encoding

One-Hot Encoding converts category and sentiment labels into separate binary columns. Missing labels are marked as `Unknown`.

```python
category_values = df_week6["category"].fillna("Unknown").astype(str)
sentiment_values = df_week6["sentiment"].fillna("Unknown").astype(str)

category_encoded = pd.get_dummies(
    category_values, prefix="Category", dtype=int
)
sentiment_encoded = pd.get_dummies(
    sentiment_values, prefix="Sentiment", dtype=int
)

df_week6 = pd.concat(
    [df_week6, category_encoded, sentiment_encoded], axis=1
)

print("One-Hot Encoding completed.")
print("New One-Hot columns:", len(category_encoded.columns) + len(sentiment_encoded.columns))
display(df_week6[list(category_encoded.columns) + list(sentiment_encoded.columns)].head())
```

## Step 11: Label Encoding

Label Encoding assigns a numerical label to each language category.

```python
label_encoder = LabelEncoder()

df_week6["Language_Encoded"] = label_encoder.fit_transform(
    df_week6["language"].fillna("Unknown").astype(str)
)

print("Language label mapping:")
print(dict(zip(
    label_encoder.classes_,
    label_encoder.transform(label_encoder.classes_)
)))

display(df_week6[["language", "Language_Encoded"]].head(10))
```

## Step 12: Min-Max Normalization

Min-Max Scaling converts each selected numerical feature to a range between 0 and 1.

```python
minmax_scaler = MinMaxScaler()
minmax_values = minmax_scaler.fit_transform(X)

minmax_columns = [col + "_MinMax" for col in feature_columns]
df_week6[minmax_columns] = minmax_values

print("Min-Max normalized values:")
display(df_week6[minmax_columns].head())
print("\nMinimum values:")
print(df_week6[minmax_columns].min().round(2))
print("\nMaximum values:")
print(df_week6[minmax_columns].max().round(2))
```

## Step 13: Standardization

StandardScaler transforms each selected feature to have approximately zero mean and unit standard deviation.

```python
standard_scaler = StandardScaler()
standard_values = standard_scaler.fit_transform(X)

standard_columns = [col + "_Standard" for col in feature_columns]
df_week6[standard_columns] = standard_values

print("Standardized values:")
display(df_week6[standard_columns].head())
print("\nMeans after standardization:")
print(df_week6[standard_columns].mean().round(2))
```

## Step 14: Compare before and after scaling

The two histograms show the original content length and its Min-Max normalized values.

```python
plt.figure(figsize=(9, 5))
sns.histplot(X["Content_Length"], label="Before Scaling",
             kde=True, stat="density", element="step")
sns.histplot(df_week6["Content_Length_MinMax"], label="After Min-Max Scaling",
             kde=True, stat="density", element="step")
plt.title("Content Length: Before and After Min-Max Scaling")
plt.xlabel("Value")
plt.legend()
plt.tight_layout()
plt.show()
```

## Step 15: Final feature engineering report

Display the dataset dimensions, missing-value count and summary of the transformed numerical features.

```python
print("WEEK 6 - FEATURE ENGINEERING REPORT")
print("===================================")
print("Original dataset shape:", df.shape)
print("Final dataset shape:", df_week6.shape)
print("Total missing values:", df_week6.isnull().sum().sum())

print("\nSummary of normalized features:")
display(df_week6[minmax_columns].describe().round(3))

print("\nWeek 6 feature engineering completed successfully.")

# Save the transformed dataset for later use.
df_week6.to_csv("Week6_Feature_Engineered_Data.csv", index=False)
print("Saved: Week6_Feature_Engineered_Data.csv")
```

# Week 7 – Dimensionality Reduction




## Step 1: Import libraries

```python
from pathlib import Path
import io
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
from sklearn.feature_selection import VarianceThreshold
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
sns.set_theme(style="whitegrid")
```

## Step 2: Load the cleaned dataset

```python
csv_path = Path("NewsData_Renewable_Energy_Final.csv")
if csv_path.exists():
    df = pd.read_csv(csv_path)
else:
    from google.colab import files
    uploaded = files.upload()
    filename = next(iter(uploaded))
    df = pd.read_csv(io.BytesIO(uploaded[filename]))
print("Dataset shape:", df.shape)
display(df[["article_id", "Year", "Title_Length", "Description_Length", "Content_Length", "Keyword_Count"]].head())
```

## Step 3: Select existing numerical features
Use the six numerical columns already present in the cleaned dataset. Article IDs, URLs and raw text are excluded because they are not continuous measurements. Sentiment is used only to color the final plot.

Validate the input without performing additional cleaning or feature engineering.

```python
feature_columns = ["source_priority", "Year", "Title_Length", "Description_Length", "Content_Length", "Keyword_Count"]
X = df[feature_columns].copy()
if not np.isfinite(X.to_numpy(dtype=float)).all():
    raise ValueError("The numerical input contains missing or infinite values. Please check the cleaned CSV.")
print("Numerical features:", feature_columns)
display(X.describe().round(2))
```

## Step 4: Variance thresholding
Remove constant columns using a variance threshold of zero. A constant feature provides no information for distinguishing articles. This threshold is not affected by the units of measurement.

```python
selector = VarianceThreshold(threshold=0.0)
selector.fit(X)
display(pd.DataFrame({"Feature": X.columns, "Variance": selector.variances_, "Retained": selector.get_support()}))
X_var = X.loc[:, selector.get_support()].copy()
print("Constant features removed:", X.columns[~selector.get_support()].tolist())
```

## Step 5: Correlation-based feature selection
Remove a feature if its absolute Pearson correlation with an already retained feature is greater than 0.90. This limits multicollinearity. The column order determines which feature is kept from a highly correlated pair. If no pair exceeds the threshold, keep all nonconstant features.

```python
corr = X_var.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0, vmin=-1, vmax=1)
plt.title("Correlation of candidate features")
plt.tight_layout()
plt.show()
selected, removed = [], []
for column in X_var.columns:
    if any(abs(corr.loc[column, kept]) > 0.90 for kept in selected):
        removed.append(column)
    else:
        selected.append(column)
X_selected = X_var[selected]
print("Correlated features removed:", removed)
print("Selected features:", selected)
print("Feature counts:", X.shape[1], "→", X_var.shape[1], "→", X_selected.shape[1])
```

## Step 6: Standardize features and apply PCA
Standardization is required for this PCA because source priority and text lengths have different numerical scales. PCA creates uncorrelated components from the standardized features. Fit all components first and determine the minimum number needed to retain at least 95% of variance.

This is exploratory analysis of the full dataset. For predictive modeling, fit scaling, selection and PCA on the training set only.

```python
if X_selected.shape[1] < 2:
    raise ValueError("At least two retained features are needed for the PCA plot.")
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_selected)
pca = PCA(svd_solver="full")
scores = pca.fit_transform(X_scaled)
ratio = pca.explained_variance_ratio_
cumulative = np.cumsum(ratio)
k = int(np.searchsorted(cumulative, 0.95) + 1)
variance = pd.DataFrame({"Component": [f"PC{i+1}" for i in range(len(ratio))], "Explained variance (%)": ratio * 100, "Cumulative variance (%)": cumulative * 100})
display(variance.round(2))
print("Components needed for at least 95% variance:", k)
```

## Step 7: Visualize individual and cumulative explained variance

```python
component_numbers = np.arange(1, len(ratio) + 1)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].bar(component_numbers, ratio * 100, color="teal")
axes[0].set(title="Scree plot", xlabel="Principal component", ylabel="Explained variance (%)")
axes[1].plot(component_numbers, cumulative * 100, marker="o", color="purple")
axes[1].axhline(95, linestyle="--", color="red", label="95% threshold")
axes[1].axvline(k, linestyle=":", color="grey", label=f"{k} components")
axes[1].set(title="Cumulative explained variance", xlabel="Number of components", ylabel="Cumulative variance (%)", ylim=(0, 105))
for ax in axes:
    ax.set_xticks(component_numbers)
axes[1].legend()
plt.tight_layout()
plt.show()
```

## Step 8: Two-dimensional PCA visualization
Plot PC1 and PC2 and color the articles by their existing sentiment labels. Sentiment is not an input to PCA. The percentage printed below shows how much information the two-component view retains; it may be less than 95%.

```python
plot_data = pd.DataFrame(scores[:, :2], columns=["PC1", "PC2"])
plot_data["Sentiment"] = df["sentiment"].fillna("unknown").str.lower()
plt.figure(figsize=(9, 6))
sns.scatterplot(data=plot_data, x="PC1", y="PC2", hue="Sentiment", alpha=0.45, s=25)
plt.xlabel(f"PC1 ({ratio[0]:.1%} variance)")
plt.ylabel(f"PC2 ({ratio[1]:.1%} variance)")
plt.title("Renewable-energy articles: PCA projection")
plt.tight_layout()
plt.show()
print(f"First two components retain {cumulative[1]:.2%} of the selected-feature variance.")
```

## Step 9: Interpret component weights
The coefficients show each feature's contribution to PC1 and PC2. Larger absolute values indicate stronger contributions. These are mathematical associations, not causal effects.

```python
weights = pd.DataFrame(pca.components_[:2].T, index=X_selected.columns, columns=["PC1", "PC2"])
display(weights.round(3))
weights.plot.barh(figsize=(9, 6))
plt.title("Feature weights in the first two components")
plt.xlabel("Component coefficient")
plt.tight_layout()
plt.show()
for pc in weights.columns:
    top = weights[pc].abs().nlargest(3).index.tolist()
    print(pc, "largest absolute weights:", ", ".join(top))
```

## Step 10: Results and reduced dataset

```python
reduced = pd.DataFrame(scores[:, :k], columns=[f"PC{i+1}" for i in range(k)])
reduced.insert(0, "article_id", df["article_id"].to_numpy())
reduced["sentiment"] = df["sentiment"].to_numpy()
reduced.to_csv("Week7_PCA_Reduced_Data.csv", index=False)
variance.to_csv("Week7_Explained_Variance.csv", index=False)
print(f"Articles analyzed: {len(df):,}")
print(f"Original numerical features: {X.shape[1]}")
print(f"Features after selection: {len(selected)}")
print(f"Components for 95% variance: {k}")
print(f"Variance retained: {cumulative[k-1]:.2%}")
if k == len(selected):
    print("All selected dimensions are needed for the 95% target.")
else:
    print(f"PCA reduced {len(selected)} selected dimensions to {k} components.")
print("CSV outputs saved in the runtime folder.")
display(reduced.head())
assert np.isfinite(scores).all()
assert np.isclose(ratio.sum(), 1)
assert cumulative[k-1] >= 0.95 - 1e-12
assert "sentiment" not in X_selected.columns
print("Validation checks passed.")
```

## Conclusion
Variance thresholding identifies constant numerical features, and correlation filtering checks for redundant features. PCA summarizes the selected standardized features. The scree plot and cumulative variance curve show how many components are required for the 95% target; the results above report the actual values.

The analysis describes news article lengths, keyword counts and source metadata. It does not measure renewable-energy generation or weather. Source priority is a supplied metadata value, not verified source quality. Large values may influence PCA even after standardization; the existing cleaned numerical values are used as supplied.
