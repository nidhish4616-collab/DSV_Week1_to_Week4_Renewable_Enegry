##WEEK 3 – Missing Values and Outlier Detection

Step 1: Missing Value Detection And Total Missing Values

```python
print("Missing Values:")
print(df.isnull().sum())

print("Total Missing Values:", df.isnull().sum().sum())
```

Step 2: Missing Value Percentage

```python
missing_percent = (df.isnull().sum() / len(df)) * 100
display(missing_percent.to_frame("Missing Percentage"))
```

Step 3: Missing Values Heatmap

```python
if df.shape[0] > 0 and df.shape[1] > 0:

    plt.figure(figsize=(14,6))

    sns.heatmap(df.isnull(), cbar=False, cmap='viridis')

    plt.title("Missing Values Visualization")
    plt.xlabel("Columns")
    plt.ylabel("Rows")

    plt.show()

else:
    print("Cannot generate heatmap: Dataset is empty.")
```

Step 4: Remove Completely Empty Column

```python
# Remove completely empty columns
df.dropna(axis=1, how='all', inplace=True)

# Numerical missing values
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Categorical missing values
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    df[col] = df[col].fillna("Unknown")

# Remove duplicates
df.drop_duplicates(inplace=True)

print("Total Missing Values:", df.isnull().sum().sum())
```

Step 5: Z-Score Outlier Detection

```python
from scipy.stats import zscore

numeric_cols = df.select_dtypes(include=np.number).columns

z = np.abs(zscore(df[numeric_cols], nan_policy='omit'))

print("Z-score outliers (|Z| > 3):")

display(pd.Series((z > 3).sum(axis=0), index=numeric_cols))
```

Step 6: IQR Outlier Detection

```python
Q1 = df[numeric_cols].quantile(0.25)

Q3 = df[numeric_cols].quantile(0.75)

IQR = Q3 - Q1

outliers = ((df[numeric_cols] < Q1 - 1.5 * IQR) |
            (df[numeric_cols] > Q3 + 1.5 * IQR))

print("IQR Outliers:")

display(outliers.sum())
```

Step 7: IQR Outlier Detection

```python
plt.figure(figsize=(10,5))

sns.boxplot(data=df[numeric_cols])

plt.title("Outlier Detection using Boxplot")

plt.show()
```

##WEEK 4 – Univariate and Bivariate EDA

Step 1: Distribution of News Categories

```python
plt.figure(figsize=(10,5))

sns.countplot(data=df, x='category',
              order=df['category'].value_counts().index)

plt.title("Distribution of News Categories")

plt.xticks(rotation=45)

plt.show()
```

Step 2: News Sentiment Distribution

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='sentiment',
              order=df['sentiment'].value_counts().index)

plt.title("News Sentiment Distribution")

plt.xlabel("Sentiment")

plt.ylabel("Number of Articles")

plt.show()
```

Step 3: Top 10 News Sources

```python
plt.figure(figsize=(12,6))

top_sources = df['source_id'].value_counts().head(10)

sns.barplot(x=top_sources.values, y=top_sources.index)

plt.title("Top 10 News Sources")

plt.xlabel("Number of Articles")

plt.ylabel("News Sources")

plt.show()
```

Step 4: Top 10 Countries

```python
plt.figure(figsize=(12,6))

top_countries = df['country'].value_counts().head(10)

sns.barplot(x=top_countries.values, y=top_countries.index)

plt.title("Top 10 Countries by News Articles")

plt.xlabel("Number of Articles")

plt.ylabel("Country")

plt.show()
```

Step 5: News Category vs Sentiment

```python
plt.figure(figsize=(12,6))

sns.countplot(data=df, x='category', hue='sentiment')

plt.title("News Category vs Sentiment")

plt.xticks(rotation=45)

plt.show()
```

Step 6: News Articles Published Per Year

```python
df['pubDate'] = pd.to_datetime(df['pubDate'], errors='coerce')

df['Year'] = df['pubDate'].dt.year

year_count = df['Year'].value_counts().sort_index()

plt.figure(figsize=(10,5))

sns.lineplot(x=year_count.index, y=year_count.values, marker='o')

plt.title("News Articles Published Per Year")

plt.xlabel("Year")

plt.ylabel("Number of Articles")

plt.show()
```

Step 7: News Category Distribution – Pie Chart

```python
category_count = df['category'].value_counts()

plt.figure(figsize=(8,8))

plt.pie(category_count.values,
        labels=category_count.index,
        autopct='%1.1f%%')

plt.title("News Category Distribution")

plt.show()
```

Step 8:Pairplot

```python
sns.pairplot(
    df[numeric_cols].sample(min(500, len(df)), random_state=42),
    diag_kind='hist'
)

plt.suptitle("Pairplot of News Dataset", y=1.02)
plt.show()
```

Step 9: Correlation Heatmap

```python
# Create numerical features
df['Title_Length'] = df['title'].fillna('').str.len()
df['Description_Length'] = df['description'].fillna('').str.len()
df['Content_Length'] = df['content'].fillna('').str.len()
df['Keyword_Count'] = df['keywords'].fillna('').str.count(',') + 1

# Select numerical columns
numeric_cols = [
    'source_priority',
    'Title_Length',
    'Description_Length',
    'Content_Length',
    'Keyword_Count'
]

# Correlation matrix
corr = df[numeric_cols].corr()

# Heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap of News Dataset")
plt.show()
```
