## Week 3 — Cleaning Sprint

```python
# The dataset uses -999 as a missing-value code
df = df.replace(-999, np.nan)

print("Missing values after converting -999 to NaN:")
print(df.isnull().sum())
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100
display(missing_percent.to_frame("Missing Percentage"))
```

```python
plt.figure(figsize=(12,5))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing Values Visualization")
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

print("Missing values after median imputation:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(14,6))
sns.boxplot(data=df[numeric_cols])
plt.xticks(rotation=90)
plt.title("Outlier Detection using Boxplot")
plt.show()
```

```python
z = np.abs(zscore(df[numeric_cols]))
print("Z-score outliers (|Z| > 3):")
display(pd.Series((z > 3).sum(axis=0), index=numeric_cols))
```

```python
Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)
IQR = Q3 - Q1

outliers = ((df[numeric_cols] < Q1 - 1.5*IQR) |
            (df[numeric_cols] > Q3 + 1.5*IQR))

print("IQR outliers:")
display(outliers.sum())
```

## Week 4 — EDA Deep Dive

```python
# Univariate analysis
column = "Solar_Irradiance_Wm2"

plt.figure(figsize=(8,5))
sns.histplot(df[column], bins=30, kde=True)
plt.title("Distribution of " + column)
plt.xlabel(column)
plt.ylabel("Frequency")
plt.show()
```

```python
# Boxplot
column = "Solar_Irradiance_Wm2"

plt.figure(figsize=(8,5))
sns.boxplot(x=df[column])
plt.title("Boxplot of " + column)
plt.xlabel(column)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Wind_Speed_kmh"], bins=30, kde=True)
plt.title("Distribution of Wind_Speed_kmh")
plt.xlabel("Wind_Speed_kmh")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Temperature_C", y="Solar_Irradiance_Wm2")
plt.title("Temperature vs Solar Irradiance")
plt.xlabel("Temperature_C")
plt.ylabel("Solar_Irradiance_Wm2")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Humidity_pct", y="Solar_Irradiance_Wm2")
plt.title("Humidity vs Solar Irradiance")
plt.xlabel("Humidity_pct")
plt.ylabel("Solar_Irradiance_Wm2")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Wind_Speed_kmh", y="Total_Renewable_Energy_kWh")
plt.title("Wind Speed vs Total Renewable Energy")
plt.xlabel("Wind_Speed_kmh")
plt.ylabel("Total_Renewable_Energy_kWh")
plt.show()
```

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
pair_cols = ['Temperature_C', 'Humidity_pct', 'Wind_Speed_kmh', 'Solar_Irradiance_Wm2', 'Cloud_Cover_pct']
sns.pairplot(df[pair_cols].sample(min(500, len(df)), random_state=42))
plt.show()
```
