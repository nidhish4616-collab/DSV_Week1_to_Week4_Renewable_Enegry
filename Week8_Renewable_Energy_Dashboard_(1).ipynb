## Step 3 – Check dataset information

The dataset contains article details such as title, country, category, publication date, sentiment and content length.

```python
print("Dataset shape:", df.shape)
print("\\nColumn names:")
print(df.columns.tolist())

print("\\nMissing values:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\\nData types:")
display(df.dtypes.to_frame("Data Type"))
```

## Step 4 – Create interactive dashboard

Use the dropdown filters to select a year, country, sentiment and category. Choose **All** to include every value.

```python
# Filter controls
year_options = ["All"] + sorted(df["Year"].dropna().astype(int).unique().tolist())
country_options = ["All"] + sorted(df["country"].dropna().astype(str).unique().tolist())
sentiment_options = ["All"] + sorted(df["sentiment"].dropna().astype(str).unique().tolist())
category_options = ["All"] + sorted(df["category"].dropna().astype(str).unique().tolist())

year_filter = widgets.Dropdown(options=year_options, value="All", description="Year:")
country_filter = widgets.Dropdown(options=country_options, value="All", description="Country:")
sentiment_filter = widgets.Dropdown(options=sentiment_options, value="All", description="Sentiment:")
category_filter = widgets.Dropdown(options=category_options, value="All", description="Category:")

run_button = widgets.Button(description="Update Dashboard", button_style="success")
output = widgets.Output()

display(widgets.VBox([
    widgets.HBox([year_filter, country_filter]),
    widgets.HBox([sentiment_filter, category_filter]),
    run_button,
    output
]))
```

```python
def create_dashboard(_=None):
    with output:
        clear_output(wait=True)

        data = df.copy()

        if year_filter.value != "All":
            data = data[data["Year"] == int(year_filter.value)]
        if country_filter.value != "All":
            data = data[data["country"].astype(str) == country_filter.value]
        if sentiment_filter.value != "All":
            data = data[data["sentiment"].astype(str) == sentiment_filter.value]
        if category_filter.value != "All":
            data = data[data["category"].astype(str) == category_filter.value]

        if data.empty:
            print("No records match the selected filters.")
            return

        # Key statistics
        total_articles = len(data)
        total_countries = data["country"].nunique()
        total_categories = data["category"].nunique()
        positive_articles = (data["sentiment"].astype(str).str.lower() == "positive").sum()

        print("RENEWABLE ENERGY NEWS – KEY STATISTICS")
        display(pd.DataFrame({
            "Metric": ["Total Articles", "Countries", "Categories", "Positive Articles"],
            "Value": [total_articles, total_countries, total_categories, positive_articles]
        }))

        # 1. Articles by country
        country_data = data["country"].value_counts().head(10).reset_index()
        country_data.columns = ["Country", "Articles"]
        fig1 = px.bar(country_data, x="Country", y="Articles",
                      title="1. Top 10 Countries by News Articles",
                      color="Articles")
        fig1.show()

        # 2. Sentiment distribution
        sentiment_data = data["sentiment"].value_counts().reset_index()
        sentiment_data.columns = ["Sentiment", "Articles"]
        fig2 = px.pie(sentiment_data, names="Sentiment", values="Articles",
                      title="2. News Sentiment Distribution", hole=0.35)
        fig2.show()

        # 3. Articles by category
        category_data = data["category"].value_counts().head(10).reset_index()
        category_data.columns = ["Category", "Articles"]
        fig3 = px.bar(category_data, x="Category", y="Articles",
                      title="3. Top News Categories", color="Category")
        fig3.show()

        # 4. Monthly publication trend
        dated = data.dropna(subset=["pubDate"]).copy()
        if not dated.empty:
            monthly = (dated.set_index("pubDate").resample("ME").size()
                       .reset_index(name="Articles"))
            fig4 = px.line(monthly, x="pubDate", y="Articles", markers=True,
                           title="4. Monthly News Publication Trend")
            fig4.show()

        # 5. Country vs sentiment heatmap
        heat = pd.crosstab(data["country"], data["sentiment"])
        top_countries = data["country"].value_counts().head(10).index
        heat = heat.loc[heat.index.intersection(top_countries)]
        if not heat.empty:
            fig5 = px.imshow(heat, text_auto=True, aspect="auto",
                             title="5. Country vs Sentiment Heatmap",
                             color_continuous_scale="Blues")
            fig5.show()

        # 6. Article content length distribution
        if "Content_Length" in data.columns:
            content_data = data.dropna(subset=["Content_Length"])
            fig6 = px.histogram(content_data, x="Content_Length", nbins=30,
                                title="6. Article Content Length Distribution")
            fig6.show()

        # Automatically generated observations
        print("KEY INSIGHTS")
        print("Most covered country:", data["country"].value_counts().idxmax())
        print("Most common category:", data["category"].value_counts().idxmax())
        print("Most common sentiment:", data["sentiment"].value_counts().idxmax())

        print("\\nSTORY")
        print("CONTEXT: The dashboard explores global renewable energy news coverage.")
        print("CONFLICT: Sentiment and coverage vary across countries and categories.")
        print("RESOLUTION: The charts help identify the most covered countries, popular topics and dominant sentiment.")

        print("\\nFiltered article records:")
        display(data[["title", "country", "category", "sentiment", "pubDate"]].head(20))

run_button.on_click(create_dashboard)

# Display the dashboard once using the default filters
create_dashboard()
```
